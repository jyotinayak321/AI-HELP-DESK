# ***K-1 OneHotEncoding***

# **What is the Dummy Variable Trap?**

When all dummy variables are kept, one dummy column can be predicted from the others.

For example:

fuel_CNG + fuel_Diesel + fuel_Petrol = 1

If you know any two values, you can always determine the third. This creates perfect multicollinearity, which is called the Dummy Variable Trap.

This can cause problems for models such as Linear Regression and Logistic Regression.

```python
#drop_first=True removes the first category from each categorical column after performing One-Hot Encoding
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True)
```

K-1 Encoding is a variation of One-Hot Encoding where, instead of creating K binary columns for K categories, we create only K-1 columns by dropping one category. The dropped category acts as the reference category. This helps avoid the Dummy Variable Trap (multicollinearity), especially in linear models like Linear Regression, while preserving all the information.

During machine learning:

Training Data → Use fit_transform()
Learns all the unique categories.
Converts them into numerical columns.

Test Data → Use only transform()

Does not learn anything new.
Uses the same categories learned from the training data.
