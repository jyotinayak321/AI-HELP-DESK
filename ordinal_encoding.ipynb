OrdinalEncoder is used to convert categorical (text) data into numerical values.

Machine learning algorithms cannot directly understand text like:

Small
Medium
Large

So we convert them into numbers

```python
from sklearn.model_selection import train_test_split
```

# **Create Encoder**

```python
oe = OrdinalEncoder(
    categories=[
        ['Poor','Average','Good'],  #specify categories Because these values have an order.
        ['School','UG','PG']
    ]
)
```
